# Contact Book for Beginners friendly:

```python
import json
contacts =[]
def save_contacts():
    with open("contacts.json", "w") as file:
        json.dump(contacts, file, indent=4)
        print('contact saved successfully.')

def load_contacts():
    global contacts

    try:
        with open("contacts.json", "r") as file:
            contacts = json.load(file)
            print("Contacts loaded successfully.")
    except (FileNotFoundError, json.JSONDecodeError):
        contacts = []
def add_contact():
    name = input("Enter contact name:")
    for contact in contacts:
        if contact['name'].lower()== name.lower():
            print("Contact with that name already exists.")
            return
    contact = {
        "name" : name,
        "email" :input("Enter your Email Address:"),
        "phone_number":input("Enter your Phone Number:")
    }

    contacts.append(contact)
    save_contacts()
    print("Contact added successfully!")

def view_contact():
    if not contacts:
        print("No contacts added!")


    else:
        for contact in contacts:
            print('\n---CONTACTS----')
            print(f"Name:{contact['name']}")
            print(f"Email:{contact['email']}")
            print(f"Phone Number:{contact['phone_number']}")
            print("-" * 25)

def delete_contact():
    if not contacts:
        print('No contacts added yet!')
        return

    else:
        name= input("Enter contact name to delete:")

        for contact in contacts:
            if contact['name'].lower() == name.lower():
                contacts.remove(contact)
                save_contacts()
                print("Contact deleted successfully!")
                return
        print("Contact not found!")


def update_contact():
    if not contacts:
        print('No contacts yet.')
        return
    name = input('Enter the name of the contact to update:')

    for contact in contacts:
        if contact['name'].lower() == name.lower():

            new_name = input(f"new name: {contact['name']}")
            new_email = input(f"new email:{contact['email']}")
            new_phone_number = input(f"new Phone Number: {contact['phone_number']}")

            if new_name:
                contact['name'] = new_name

            if new_email:
                contact['email'] = new_email
            if new_phone_number:
                contact['phone_number'] = new_phone_number
            save_contacts()
            print("Contact updated successfully!")
            return

    print("Contact not found.")

def search_contact():
    if not contacts:
        print("No contacts found.")
        return

    name= input("Enter the name of the contact to search?:")

    for contact in contacts:
        if contact['name'].lower()== name.lower():

            print(f"Name:{contact['name']}")
            print(f"Email:{contact['email']}")
            print(f"Phone Number:{contact['phone_number']}")
            return

    print("Contact not found!")

def menu():
    while True:
        print("\n-----CONTACT BOOK-----")
        print("1. Add contact.")
        print("2. View contact.")
        print("3. delete contact.")
        print("4. Update contact.")
        print("5. Search contact.")
        print("6. Exit.")
        choice= input("Enter your choice:")

        if choice == '1':
            add_contact()
        elif choice== '2':
            view_contact()
        elif choice== '3':
            delete_contact()
        elif choice== '4':
            update_contact()

        elif choice =="5":
            search_contact()



        elif choice== '6':
            print("Thank you, for using this contact book.")
            break
        else:
            print("Invalid choice!")

load_contacts()
menu()

```
